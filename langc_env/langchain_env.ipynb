```python
import os
from google.colab import userdata
from langchain.chat_models import init_chat_model
from langchain.messages import HumanMessage, SystemMessage
from IPython.display import HTML, display

# Enable line wrapping dynamically
def enable_line_wrap():
    display(HTML('''
    <style>
      pre {
          white-space: pre-wrap !important;
          word-break: break-all !important;
      }
      .output_pre {
          white-space: pre-wrap !important;
          word-break: break-all !important;
      }
    </style>
    '''))

enable_line_wrap()

# Retrieve the OpenAI API key securely from Colab Secrets
try:
    os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
except Exception as e:
    print("Please add your OPENAI_API_KEY to the Colab Secrets (key icon on the left) and enable notebook access.")

model = init_chat_model(
    os.getenv("OPENAI_MODEL", "gpt-4o-mini"),
    model_provider="openai",
)

messages = [
    SystemMessage("You are an AI tutor, Explain concepts intuitively first."),
    HumanMessage("Teach me cosine similarity"),
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

enable_line_wrap()

model = ChatOpenAI(
    model=os.getenv("OPENAI_MODEL", "gpt-5.5"),
    temperature=0,
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a {role}. Answer for a {audiance} audiance."),
    ("human", "{question}")
])

messages = prompt.invoke({
    "role": "Machine learning professor",
    "audiance": "beginner engineering",
    "question": "Explain attention."
})

chain = prompt | model | StrOutputParser()

answer = chain.invoke({
    "role": "Machine learning professor",
    "audiance": "beginner engineering",
    "question": "Explain attention."
})

print(answer)
```
