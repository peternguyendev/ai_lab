```python
from pydantic import BaseModel, Field

class ProductReview(BaseModel):
  sentiment: str = Field(description="positive, neutral, or negative")
  score: int = Field(ge=1, le=5)
  summary: str

structured_model = model.with_structured_output(ProductReview)

review = structured_model.invoke(
    "The camera is excellent, but battery life is mediocre. 4/5."
)

print(review.sentiment)
print(review.score)
print(review.summary)
```

```python
from langchain.tools import tool

@tool
def resistor_parallel(r1: float, r2: float) -> float:
  """Calculate the equivalent resitance of two resistors in pparallel."""
  if r1 <= 0 or r2 <= 0:
    raise ValueError("Resistance must be possitive.")
  return (r1 * r2) / (r1 + r2)

print(resistor_parallel.invoke({"r1": 1000, "r2": 2000}))
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
from google.colab import userdata
import os

# Securely load the OpenAI API key
try:
    os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
except Exception as e:
    print("Please add your OPENAI_API_KEY to the Colab Secrets (key icon on the left) and enable notebook access.")

# Short term memory: Checkpointer saves thread-level state for later retrieval
agent = create_agent(
    model = f"openai:{os.getenv('OPENAI_MODEL', 'gpt-5.5')}",
    tools=[],
    checkpointer=InMemorySaver(),
)
config = {"configurable": {"thread_id": "user-42"}}

# Provide model with information and save for later use.
agent.invoke(
    {"messages": [{"role": "user", "content": "My project is called Orion."}]},
    config,
)

# Retrieve saved checkpoint
result = agent.invoke(
    {"messages": [{"role": "user", "content": "What is my project called?"}]},
    config,
)

print(result["messages"][-1].content)
```

```python
# Streaming: improve latency due to awaiting for long output to finish.
# Agent will wait for the content available then print and flush, wait for new # content.

import os
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model=os.getenv("OPENAI_MODEL", "gpt-5.5"),
    temperature=0,
)

for chunk in model.stream("Explain transformers in simple terms"):
  if chunk.content:
    print(chunk.content, end="", flush=True)
```

```python
# 2-step RAG
import os
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. Source documents
documents = [
    Document(
        page_content=(
            "LangChain is a framework for building LLM application. "
            "It includes models, tools, agents, retrieval, and middleware."
        )
    ),
    Document(
        page_content=(
            "LangGraph is a lower-level orchestration framework used for "
            "stateful and durable agent workflows. "
        )
    ),
    Document(
        page_content=(
            "RAG retrieves relevant external information at query time "
            "and provides it to a language model as context. "
        )
    ),
]

# 2. Embed + index (Fixed typo 'Text-embedding-3-samll' to 'text-embedding-3-small')
embedding_model = OpenAIEmbeddings(model="text-embedding-3-small")
vectorstore = FAISS.from_documents(documents, embedding_model)
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})

# 3. Model
# Securely load the OpenAI API key
try:
    os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
except Exception as e:
    print("Please add your OPENAI_API_KEY to the Colab Secrets (key icon on the left) and enable notebook access.")

model = ChatOpenAI(
    model=os.getenv("OPENAI_MODEL", "gpt-5.5"),
    temperature=0,
)

# 4. Prompt
prompt = ChatPromptTemplate.from_template("""
You are a technical tutor. Answer the question using only the supplied context.
If the context does not contain the answer, say that you do not know.

Context:
{context}

Question
{question}
""")

# 5. Helper to format retrieved documents (Fixed unclosed quote in join statement)
def format_docs(docs):
  return "\n\n".join(doc.page_content for doc in docs)

# 6. Retrieve first
question = "What is the relationship between LangChain and LangGraph?"
docs = retriever.invoke(question)
context = format_docs(docs)

# 7. Generate
chain = prompt | model | StrOutputParser()
answer = chain.invoke({
    "context": context,
    "question": question,
})

print(answer)
```

```python
# Agentic RAG
import os
from google.colab import userdata
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
from IPython.display import HTML, display

# Securely load the OpenAI API key
try:
    os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
except Exception as e:
    print("Please add your OPENAI_API_KEY to the Colab Secrets (key icon on the left) and enable notebook access.")

# Define the documents that were lost on kernel restart
documents = [
    Document(
        page_content=(
            "LangChain is a framework for building LLM application. "
            "It includes models, tools, agents, retrieval, and middleware."
        )
    ),
    Document(
        page_content=(
            "LangGraph is a lower-level orchestration framework used for "
            "stateful and durable agent workflows. "
        )
    ),
    Document(
        page_content=(
            "RAG retrieves relevant external information at query time "
            "and provides it to a language model as context. "
        )
    ),
]

embedding_model = OpenAIEmbeddings(model="text-embedding-3-small")
vectorstore = FAISS.from_documents(documents, embedding_model)
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})

# Enable line wrapping dynamically
def enable_line_wrap():
    display(HTML('''
    <style>
      pre {
          white-space: pre-wrap !important;
          word-break: break-all !important;
      }
      .output_pre {
          white-space: pre-wrap !important;
          word-break: break-all !important;
      }
    </style>
    '''))

@tool
def search_knowledge_base(query: str) -> str:
  """Search the internal knowledge base for information about LangChain."""
  docs = retriever.invoke(query)
  return "\n\n".join(d.page_content for d in docs)

agent = create_agent(
    model = f"openai:{os.getenv('OPENAI_MODEL', 'gpt-5.5')}",
    tools=[search_knowledge_base],
    system_prompt=(
        "You are a technical assistant. "
        "Use search_knowledge_base for questions about our internal knowledge. "
    ),
)

# Corrected the input key from "message" to "messages"
result = agent.invoke({
    "messages": [{
        "role": "user",
        "content": "Based on the knowledge base, what is RAG? "
    }]
})

# Corrected from slicing a list [:-1] to targeting the last response message object [-1]
enable_line_wrap()
print(result["messages"][-1].content)
```
